
### Ejercicio con Geocodificación y visualización en Google My Maps

**Objetivo:** Aprender a geocodificar direcciones con `geopy`, generar un archivo `.kml` y visualizarlo en Google My Maps.

#### Instrucciones:

1. Crea un archivo llamado `direcciones.txt` con al menos 5 direcciones reales de México:

   Ejemplo:  
   ```
   Calle Madero 45;Centro;Ciudad de México;06000
   Av. Universidad 3000;Coyoacán;Ciudad de México;04510
   ```

2. Carga el archivo con `pandas` utilizando `;` como delimitador.

3. Crea una columna `direccion_completa` concatenando calle, colonia, ciudad y código postal.

4. Geocodifica usando `geopy` (servicio Nominatim) y obtén latitud y longitud.

5. Crea un `GeoDataFrame` con geometría tipo `Point`.

6. Exporta el resultado a un archivo `.kml` con `gdf.to_file("direcciones.kml", driver="KML")`.

7. Sube el archivo a [Google My Maps](https://www.google.com/mymaps), crea un mapa nuevo e importa el `.kml`.
