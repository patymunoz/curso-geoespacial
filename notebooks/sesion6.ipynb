#### Colores

Usa **cmap** cuando:

* Estás representando valores numéricos continuos u ordinales (como índices, tasas, densidades).

* Quieres aplicar una escala de colores graduada automáticamente.

* Estás haciendo un mapa de coropletas, un heatmap, o cualquier visualización basada en valores numéricos.

Aquí puedes consultar todos los **colormaps** disponibles en Matplotlib: [Colormaps](https://matplotlib.org/stable/users/explain/colors/colormaps.html)

## Orden de capas

El ejercio anterior lo trabajamos únicamente con una sola capa, de geometría de polígonos pero, ¿qué pasa si tenemos varias capas y queremos ordenarlas para presentarlas en un mapa?

Acá vamos a integrar una capa de puntos para analizar visualmente si las zonas de mayo marginación que estamos viendo en la capa de polígonos están relacionados con la ubicación de **hogares con población indígena**.

Esta es la forma más sencilla de superponer capas. A medida que agreguemos más capas al mapa, utilizaremos ``plot()`` para dibujarlas una a una. Para controlar el orden en que se apilan visualmente, emplearemos el parámetro ``zorder``, que nos permite definir qué capa aparece por encima o por debajo de las demás.

### Contextily

Contextily también lo hemos usado en sesiones anteriores, así que veremos cómo se integra:

Puedes consultar la documentación oficial de contextily aquí: [Contextily](https://contextily.readthedocs.io/en/latest/intro_guide.html). Para que el mapa funcione, es importante que el `CRS` de las capas sea compatible con el sistema de referencia de coordenadas utilizado por `contextily`, que es `EPSG:3857`. Si tus capas no están en este CRS, puedes reproyectarlas utilizando el método `to_crs()` de GeoPandas.
