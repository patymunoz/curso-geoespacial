## Sistema de coordenadas de referencia

OpenStreetMap utiliza el sistema de coordenadas ``WGS84``, al igual que la mayoría de unidades GPS.

Los objetos de grafo en OSMnx (como los que obtienes con ``graph_from_place``,``graph_from_polygon``, etc.) no tienen un método interno para cambiar su sistema de coordenadas (CRS) directamente.
Es decir, no puedes hacer ``grafo.to_crs(...)`` como sí puedes con un GeoDataFrame de GeoPandas.

Sin embargo, OSMnx te proporciona una función específica para este propósito, llamada:

`os.project_graph()`

Esta función toma un grafo como entrada y lo transforma a otro sistema de referencia (CRS). Si no le indicas un CRS explícito, automáticamente transforma el grafo al sistema ``UTM`` más adecuado para la ubicación de los datos (lo cual es útil y conveniente).

### 📐 ¿Cómo calcular el área cubierta por la red?

OSMnx no calcula automáticamente el área de cobertura de la red vial. Para obtenerla manualmente:

1. Convertimos el grafo a `GeoDataFrame` de aristas
2. Creamos una **unión geométrica** de todas las líneas
3. Generamos un contorno (convex hull o concave hull)
4. Calculamos el área del contorno (preferiblemente proyectado en metros)

### La ruta

Se utiliza para calcular una ruta entre dos nodos en un grafo.

**Parámetros obligatorios:**

`graph` = El grado de calles o conexiones.

`origin` = El identificador (ID) del nodo de origen en el grafo.

`destination` = El identificador (ID) del nodo de detino.

**Parámetros opcionales**

`weight` = Define el criterio de optimización de la ruta; por defecto es `lenght`, lo que signfica que se minimiza la distancia en metros pero puedes usar otro atributo como `travel_time` si se ha calculado previamente. 

`cpus` = Controla la cantidad de núcleso de CPU a usar si se están calculando muchas rutas. Útil para parelelizar el procesamiento y acelerar el cálculo.
